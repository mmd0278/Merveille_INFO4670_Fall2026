### Q1 · One quantitative variable
**How many hours per week do Northgate students work?** Describe what's typical, how spread out it is, and whether anyone stands apart.
→ *histogram* (add a *box plot* if it helps). Column: `work_hours_per_week` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax)
ax.set_title("Weekly Work Hours for Northgate Students")
ax.set_xlabel("Work Hours per Week")
ax.set_ylabel("Number of Students")
plt.show()
```

**STAR story - Q1**

**Scope:** This chart shows the number of hours per week that Northgate students work.

**Track:** I am looking at how many hours per week Northgate students typically work, how spread out the hours are, and whether any values stand out.

**Articulate:** Most students work about 10 to 25 hours per week, with many students around 15 to 20 hours. The values range from about 0 to 40 hours for most students.

**Respond:** There is an unusual negative value near -10 hours, which does not make sense for weekly work hours. I would look at that student record to see if it is a dataentry error.

### Q2 · One categorical variable
**How does enrollment compare across majors?**
→ *sorted bar chart*. Column: `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---

major_counts = students["major"].value_counts().sort_values(ascending=False)

fig, ax = plt.subplots()
sns.barplot(x=major_counts.values, y=major_counts.index, color=NAVY, ax=ax)
ax.set_title("Northgate Student Enrollment by Major")
ax.set_xlabel("Number of Students")
ax.set_ylabel("Major")
plt.show()
```

**STAR story - Q2**

- **Scope —** This chart shows the number of Northgate students enrolled in each major.
- **Track —** I am comparing enrollment across majors to see which majors have the most and least students.
- **Articulate —** Cybersecurity has the highest enrollment with more than 500 students. Data Science, Information Science, and Information Technology each have around 280 students.
- **Respond —** Some majors appear more than once with slightly different names, such as Cybersecurity and Cyber Security, and Data Science and DataScience. I would check these labels to see if they should be combined.

### Q3 · A quantity, split by category
**Does final GPA differ across majors?** Read the exceptions, not just the boxes.
→ *box plot per major*. Columns: `final_gpa` and `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---

fig, ax = plt.subplots()
sns.boxplot(data=students, x="major", y="final_gpa", color=NAVY, ax=ax)
ax.set_title("Final GPA by Major")
ax.set_xlabel("Major")
ax.set_ylabel("Final GPA")
plt.xticks(rotation=45, ha="right")
plt.show()
```

**STAR story - Q3**

- **Scope —** This chart shows the final GPA of Northgate students separated by major.
- **Track —** I am comparing final GPA across majors to see whether some majors have higher or lower GPAs than others.
- **Articulate —** Final GPA is fairly similar across the majors. Most median GPAs are around 2.2 to 2.5, and the ranges overlap considerably.
- **Respond —** I noticed several unusually low GPA values, including some near 0. I would look more closely at those students to understand why their final GPA is so low.

### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---

fig, ax = plt.subplots()

sns.regplot(
    data=students,
    x="commute_miles",
    y="final_gpa",
    scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
    line_kws={"color": RUST},
    ax=ax
)

ax.set_title("Final GPA vs. Commute Distance")
ax.set_xlabel("Commute Miles")
ax.set_ylabel("Final GPA")
plt.show()
```

**STAR story - Q4**

- **Scope —** This chart shows the relationship between commute distance and final GPA for Northgate students.
- **Track —** I am looking at whether students who commute farther tend to have different final GPAs.
- **Articulate —** The trend line slopes downward, showing a negative association between commute distance and final GPA. Students with longer commutes tend to have lower GPAs, although there is a lot of variation.
- **Respond —** I noticed several students with a final GPA of 0 and a few students with very long commutes. I would look at these students more closely and consider other factors before concluding that commute distance explains their GPA.

### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---

weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()

fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active",
             marker="o", color=NAVY, ax=ax)
ax.set_title("Northgate Student Engagement Across the Semester")
ax.set_xlabel("Week of Semester")
ax.set_ylabel("Average Minutes Active")
plt.show()
```

**STAR story - Q5**

- **Scope —** This chart shows the average weekly active minutes of Northgate students across the semester.
- **Track —** I am looking at how student engagement changes from week to week during the semester.
- **Articulate —** Student engagement stays fairly consistent at around 210 average active minutes for most weeks. However, there is a large drop in Week 9 to about 90 minutes before activity returns to its normal level in Week 10.
- **Respond —** Week 9 stands out because student activity drops much lower than every other week. I would look more closely at Week 9 to see what happened during that week.
